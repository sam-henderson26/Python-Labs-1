```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.arange(0, 4, 0.05)
y = np.sin(x*np.pi)
plt.figure(figsize=(3, 2))
plt.plot(x, y)

plt.xlabel('t [s]')
plt.ylabel('S [V]')
plt.title('Sine wave')
```

```python
# Decorating
x = np.arange(0, 4, 0.05)
y = np.sin(x*np.pi)
fig, ax = plt.subplots(
figsize = (3,2),
constrained_layout = True)
fig.set_facecolor('turquoise')
ax.plot(x, y)
ax.set_xlabel('t [s]')
ax.set_ylabel('S [V]')
ax.set_title('Sine wave')
```

```python
# Labels and Legend
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
plt.figure(figsize=(5,3))
plt.plot(x, y1, label = 'sin(x)')
plt.plot(x, y2, label = 'cos(x)')
plt.grid()
plt.legend()
plt.show()
```

```python
# Adjusting axes ranges:
x = np.linspace(-10, 10, 200)
y = x**3 - 2.0*x**2 - 1.0*x + 2.0
plt.figure(figsize=(7, 7))
plt.subplot(2, 1, 1)
plt.title('Default axes limits')
plt.plot(x, y)
plt.grid()
plt.subplot(2, 1, 2)
plt.plot(x, y)
plt.title('Custom axes limits')
plt.xlim(-5, 5)
plt.ylim(-2, 3)
plt.grid()
plt.show()
```
