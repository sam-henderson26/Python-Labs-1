```python
import numpy as np
```

```python
# Convert a list to an array
x = np.array([1, 2, 3, 4])
```

```python
# Create a range of numbers
x1 = np.arange(1, 5, 0.1) # Note: 5 (end) is excluded.
```

```python
# Create equally spaced array
x2 = np.linspace(1, 5, 20) # Note: 5 (end) is included.
```

```python
# Array of zeros
x3 = np.zeros((1, 10)) # 1 row, 10 columns
```

```python
# Array of ones
x4 = np.ones((7, 1)) # 7 rows, 1 column
```

```python
# Create a range of numbers, end point excluded
x5 = np.arange(1, 10, 1) # (start, end, step)
```

```python
# Calculate values for a mathematical expression
y1 = x + 2
```

```python
# Calculate values for another mathematical expression
y2 = x**2
```

```python
# Combining arrays, equal to: x**2 + x + 2
y3 = y2 + y1
```
