### My solution

```python
# start your code here
x = np.linspace(-10, 10, 2000)
y = ((x >= -0.9) & (x < 2.1)) * (0.9 + x) + ((x >= 2.1) & (x < 5.1)) * (5.1 - x)
plt.figure(figsize=(8, 5))
plt.plot(x, y, label='f(x)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Problem 1: Piecewise Function')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Run the code below to see your problem text
txt = mx.lab_3_problem_2(student_id)
display(Markdown(txt))
```

### My solution

```python
# Your code starts here
x = np.linspace(-10, 10, 2000)
top = (x < -5.8) * (-216.04)
middle = (x >= -5.8) * (x < 5.5) * (1.0 * x**3 - 2.04 * x**2 - 9.12 * x - 5.2)
bottom = (x >= 5.5) * (49.31)
y = top + middle + bottom
plt.figure(figsize=(8, 5))
plt.plot(x, y, label='f(x)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Problem 2: Piecewise Function')
plt.legend()
plt.grid(True)
plt.show()
```
