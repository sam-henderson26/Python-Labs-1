### My solution

```python
x = np.linspace(0, 10, 1000)
f = np.sqrt (2.4 * x)
g = (x - 5.48) ** 2
plt.figure(figsize = (7,7))
plt.plot(x, f, label='f(x)')
plt.plot(x, g, label='g(x)')
plt.title("Overview: Both functions")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize = (7,7))
plt.plot(x, f, label='f(x)')
plt.plot(x, g, label='g(x)')
plt.xlim(3.7, 3.8)
plt.ylim(2.75, 3.25)
plt.title("Intersection Point 1 (zoomed)")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

First intersection occurs at (3.75, 3)
